# Week 2: Exploratory Data Analysis (EDA) — Titanic Dataset

## Objective
Perform an end-to-end Exploratory Data Analysis on the Titanic passenger dataset by inspecting data structure and missing values, building 6 informative visualizations, and documenting clear data-driven insights.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set clean visual style
sns.set_theme(style="whitegrid", palette="Set2")

# Load built-in Titanic dataset
df = sns.load_dataset("titanic")

print("Dataset loaded successfully!")
print("Shape (Rows, Columns):", df.shape)
df.head()
```

```python
print("--- Missing Values per Column ---")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\n--- Summary Statistics (Numerical) ---")
display(df.describe().round(2))

# Clean missing 'age' values with median for accurate plotting
df["age_clean"] = df["age"].fillna(df["age"].median())
```

---
### Visualization 1: Overall Survival Distribution
**Question:** How many passengers survived versus did not survive the disaster?


```python
plt.figure(figsize=(7, 4))
ax = sns.countplot(data=df, x="survived", hue="survived", palette=["#e74c3c", "#2ecc71"], legend=False)
plt.title("Overall Passenger Survival Count (0 = No, 1 = Yes)", fontsize=13, fontweight="bold")
plt.xlabel("Survived (0 = Did Not Survive, 1 = Survived)")
plt.ylabel("Number of Passengers")

for container in ax.containers:
    ax.bar_label(container, padding=3)

plt.tight_layout()
plt.show()
```

**Insight 1:** Out of 891 passengers in the dataset, **549 (61.6%) did not survive**, while only **342 (38.4%) survived**. The dataset shows a clear class imbalance favoring non-survivors.

---
### Visualization 2: Survival Rate by Gender
**Question:** Did gender significantly influence a passenger's chance of survival?

```python
plt.figure(figsize=(7, 4))
ax = sns.barplot(data=df, x="sex", y="survived", hue="sex", estimator=lambda x: np.mean(x) * 100, errorbar=None, legend=False)
plt.title("Survival Rate (%) by Gender", fontsize=13, fontweight="bold")
plt.xlabel("Gender")
plt.ylabel("Survival Rate (%)")
plt.ylim(0, 100)

for container in ax.containers:
    ax.bar_label(container, fmt="%.1f%%", padding=3)

plt.tight_layout()
plt.show()
```

**Insight 2:** **Female passengers had a 74.2% survival rate**, compared to only **18.9% for male passengers** — nearly 4 times higher, reflecting the *"women and children first"* evacuation protocol.

---
### Visualization 3: Survival Rate by Passenger Class
**Question:** How did ticket class (socio-economic status) affect survival rates?

```python
plt.figure(figsize=(8, 4.5))
ax = sns.barplot(data=df, x="class", y="survived", hue="sex", estimator=lambda x: np.mean(x) * 100, errorbar=None)
plt.title("Survival Rate (%) by Passenger Class and Gender", fontsize=13, fontweight="bold")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate (%)")
plt.ylim(0, 110)
plt.legend(title="Gender")

for container in ax.containers:
    ax.bar_label(container, fmt="%.1f%%", padding=2, fontsize=9)

plt.tight_layout()
plt.show()
```

**Insight 3:** Higher ticket class strongly increased survival chances: **First Class passengers survived at 63.0%**, Second Class at **47.3%**, and Third Class at just **24.2%**. Even within First Class, females survived at 96.8% compared to 36.9% for males.


---
### Visualization 4: Age Distribution of Passengers
**Question:** What was the age distribution of passengers onboard, and how does it compare between survivors and non-survivors?

```python
plt.figure(figsize=(8, 4.5))
sns.histplot(data=df, x="age_clean", hue="survived", bins=25, kde=True, palette=["#e74c3c", "#2ecc71"], alpha=0.5)
plt.title("Age Distribution by Survival Status", fontsize=13, fontweight="bold")
plt.xlabel("Age (Years)")
plt.ylabel("Passenger Count")
plt.legend(title="Status", labels=["Survived", "Did Not Survive"])

plt.tight_layout()
plt.show()
```

**Insight 4:** Most passengers were young adults aged **20 to 35 years**. Notably, **young children (under age 10)** show a distinct survival spike (more survivors than non-survivors), whereas adults aged 20–30 suffered the highest total casualties.

---
### Visualization 5: Ticket Fare Distribution Across Classes
**Question:** How do ticket fares vary across passenger classes, and are there extreme outliers?

```python
plt.figure(figsize=(8, 4.5))
sns.boxplot(data=df, x="class", y="fare", hue="class", palette="Set2", legend=False)
plt.title("Ticket Fare Distribution by Passenger Class", fontsize=13, fontweight="bold")
plt.xlabel("Passenger Class")
plt.ylabel("Fare (£)")

plt.tight_layout()
plt.show()
```

**Insight 5:** First Class fares show a wide spread with a median around **£60** and extreme luxury outliers reaching over **£500**, whereas Second and Third Class tickets were tightly clustered around median fares of **£14.25** and **£8.05** respectively.

---
### Visualization 6: Correlation Heatmap of Numerical Features
**Question:** Which numerical variables correlate most strongly with survival and ticket fare?

```python
plt.figure(figsize=(8, 5))
num_cols = df[["survived", "pclass", "age_clean", "sibsp", "parch", "fare"]]
corr_matrix = num_cols.corr()

sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", linewidths=0.5, vmin=-1, vmax=1)
plt.title("Correlation Matrix of Numerical Features", fontsize=13, fontweight="bold")

plt.tight_layout()
plt.show()
```

**Insight 6:** `pclass` has the strongest negative correlation with `survived` (**-0.34**), confirming that lower-numbered classes (1st class) had much higher survival rates. `fare` correlates positively with `survived` (**+0.26**) and negatively with `pclass` (**-0.55**).

---

## Key EDA Takeaways Summary
1. **Overall Survival:** Only **38.4%** of passengers survived (342 out of 891).
2. **Primary Survival Drivers:** **Gender** (74.2% female vs. 18.9% male) and **Socio-Economic Class** (63.0% 1st class vs. 24.2% 3rd class) were the strongest predictors of survival.
3. **Age Factor:** Children under 10 had higher survival odds than young adults (ages 20–35), who made up the largest demographic onboard.
4. **Data Quality Note:** The `deck` column had 688 missing values (77% missing) and `age` had 177 missing values (19.9%), which were imputed using the median age (`28.0`) for distribution analysis.
