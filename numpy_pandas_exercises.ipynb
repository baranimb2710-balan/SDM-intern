Import Libraries

```python
import numpy as np
import pandas as pd

print("NumPy Version:", np.__version__)
print("Pandas Version:", pd.__version__)
```

1:Create a 1D Numpy array of numbers from 10 to 50

```python
arr_1d = np.arange(10, 51)
print("1D Array (10 to 50):\n", arr_1d)
```

2:Create a 3*3 identity matrix and a 4*4 with fixed randam integers

```python
identity_mat = np.eye(3)
np.random.seed(42)
random_mat = np.random.randint(1, 100, size=(4, 4))

print("3x3 Identity Matrix:\n", identity_mat)
print("\n4x4 Random Matrix:\n", random_mat)
```

3:Reshape a 1D array of 12 elements into a 3*4 2D matrix

```python
arr = np.arange(1, 13)
reshaped_arr = arr.reshape(3, 4)
print("Reshaped 3x4 Matrix:\n", reshaped_arr)
```

4: Slice the first 2 rows and last 2 columns from a 4x4 matrix.

```python
matrix_4x4 = np.arange(1, 17).reshape(4, 4)
sliced_mat = matrix_4x4[:2, -2:]

print("Original 4x4 Matrix:\n", matrix_4x4)
print("\nSliced (First 2 rows, Last 2 cols):\n", sliced_mat)
```

5:Filter all even numbers from a NumPy array using boolean indexing.

```python
nums = np.array([11, 22, 35, 44, 57, 68, 79, 80, 92])
even_nums = nums[nums % 2 == 0]
print("Even Numbers:", even_nums)
```

6: Perform element-wise addition, multiplication, and dot product of two arrays.

```python
a = np.array([2, 4, 6])
b = np.array([1, 3, 5])

print("Addition:", a + b)
print("Element-wise Multiplication:", a * b)
print("Dot Product:", np.dot(a, b))
```

7: Compute mean, median, standard deviation, min, and max of an array.

```python
scores = np.array([65, 72, 88, 91, 54, 79, 84, 95])

print("Mean:", np.mean(scores))
print("Median:", np.median(scores))
print("Standard Deviation:", round(np.std(scores), 2))
print("Min:", np.min(scores), "| Max:", np.max(scores))
```

8: Calculate row-wise sum and column-wise sum of a 2D array.

```python
data_2d = np.array([[10, 20, 30],
                    [40, 50, 60],
                    [70, 80, 90]])

print("Row-wise Sum (axis=1):", np.sum(data_2d, axis=1))
print("Column-wise Sum (axis=0):", np.sum(data_2d, axis=0))
```

9: Detect and count missing (NaN) values in a NumPy array, then replace them with 0.

```python
arr_nan = np.array([10.0, np.nan, 30.0, np.nan, 50.0])
nan_count = np.isnan(arr_nan).sum()
cleaned_arr = np.nan_to_num(arr_nan, nan=0.0)

print("Missing values count:", nan_count)
print("Cleaned Array:", cleaned_arr)
```

10: Find unique elements and sort a NumPy array.

```python
dup_arr = np.array([45, 12, 78, 12, 34, 45, 90, 34])
unique_vals = np.unique(dup_arr)
sorted_vals = np.sort(dup_arr)

print("Unique Elements:", unique_vals)
print("Sorted Array:", sorted_vals)
```

11: Create a Pandas DataFrame with student records


```python
data = {
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": ["Arun", "Priya", "Karthik", "Divya", "Sanjay", "Meera", "Vikram", "Ananya"],
    "Department": ["AI&DS", "CSE", "AI&DS", "ECE", "CSE", "AI&DS", "ECE", "CSE"],
    "Marks": [88, 92, 76, 85, np.nan, 95, 68, 81],
    "City": ["Chennai", "Madurai", "Coimbatore", "Chennai", "Trichy", "Madurai", "Chennai", "Coimbatore"]
}

df = pd.DataFrame(data)
df
```

12: Inspect the DataFrame (first 5 rows, shape, data types, and summary statistics).

```python
print("Shape of DataFrame:", df.shape)
print("\nFirst 5 Rows:\n", df.head())
print("\nSummary Statistics:\n", df.describe())
```

13: Select specific columns (Name and Marks) from the DataFrame.

```python
selected_cols = df[["Name", "Marks"]]
selected_cols
```

14: Filter rows where Marks are greater than 80.

```python
high_scorers = df[df["Marks"] > 80]
high_scorers
```

15: Check for missing values and fill missing Marks with the column mean.

```python
print("Missing values before filling:\n", df.isnull().sum())

mean_marks = round(df["Marks"].mean(), 1)
df["Marks"] = df["Marks"].fillna(mean_marks)

print("\nMissing values after filling:\n", df.isnull().sum())
df
```

16: Add a new column Result based on Marks (Pass if Marks >= 75 else Needs Improvement).

```python
df["Result"] = np.where(df["Marks"] >= 75, "Pass", "Needs Improvement")
df
```

17: Sort the DataFrame by Marks in descending order.

```python
sorted_df = df.sort_values(by="Marks", ascending=False)
sorted_df
```

18: Group by Department and calculate the average marks for each department.

```python
dept_avg = df.groupby("Department")["Marks"].mean().round(2)
print("Average Marks by Department:\n", dept_avg)
```

19: Count the number of students in each City using value_counts().

```python
city_counts = df["City"].value_counts()
print("Student Count by City:\n", city_counts)
```

20: Export the cleaned DataFrame to a CSV file and verify by reading it back.

```python
df.to_csv("cleaned_students.csv", index=False)

verified_df = pd.read_csv("cleaned_students.csv")
print("CSV saved and loaded successfully! Shape:", verified_df.shape)
verified_df.head(3)
```
