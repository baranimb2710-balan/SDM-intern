## 1. Vectors
- **Definition:** A vector is an ordered list of numbers representing a point or direction in space. In ML, a single data row (features of one house or student) is represented as a vector.
- **Example:** $\mathbf{v} = [2, 5, 8]$ (a 3-dimensional vector).
- **Key Operations:**
  - **Vector Addition:** Adding corresponding elements ($\mathbf{a} + \mathbf{b}$).
  - **Scalar Multiplication:** Scaling every element by a single number ($c \cdot \mathbf{v}$).
  - **Dot Product:** $\mathbf{a} \cdot \mathbf{b} = \sum a_i b_i$. Measures similarity/alignment between two vectors (used heavily in neural networks and cosine similarity).
  - **Magnitude (L2 Norm):** $\|\mathbf{v}\| = \sqrt{v_1^2 + v_2^2 + \dots + v_n^2}$ (length of the vector).

```python
import numpy as np


v1 = np.array([3, 4])
v2 = np.array([1, 2])

print("Vector Addition:", v1 + v2)
print("Scalar Multiplication (3 * v1):", 3 * v1)
print("Dot Product:", np.dot(v1, v2))
print("L2 Norm (Length of v1):", np.linalg.norm(v1))
```

## 2. Matrices
- **Definition:** A matrix is a 2D grid of numbers arranged in rows and columns. In ML, an entire dataset (rows = samples, columns = features) or neural network layer weights is stored as a matrix.
- **Shape:** $m \times n$ ($m$ rows, $n$ columns).
- **Key Operations:**
  - **Transpose ($A^T$):** Flipping rows into columns.
  - **Matrix Multiplication ($A \times B$):** Valid when columns of $A$ equal rows of $B$. Used to transform data and compute layer outputs ($Y = XW + b$).
  - **Identity Matrix ($I$):** A square matrix with 1s on the diagonal and 0s elsewhere ($A \cdot I = A$).

```python

X = np.array([[1, 2],
              [3, 4],
              [5, 6]])

W = np.array([[0.5, 1.0],
              [1.5, 2.0]])

print("Matrix X shape:", X.shape)
print("Transpose of X:\n", X.T)
print("\nMatrix Multiplication (X @ W):\n", X @ W)
```

## 3. Probability
- **Definition:** Probability measures the likelihood of an event occurring, ranging from $0$ (impossible) to $1$ (certain). ML classifiers output probabilities (e.g., 85% chance an email is spam).
- **Key Concepts:**
  - **Basic Probability:** $P(A) = \frac{\text{Favorable Outcomes}}{\text{Total Outcomes}}$
  - **Conditional Probability ($P(A|B)$):** Probability of event $A$ happening given that event $B$ has already occurred.
  - **Bayes' Theorem:** $P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}$ — updates our belief based on new evidence (foundation of Naive Bayes classifiers).
  - **Expectation & Variance:** Mean ($\mu$) shows the center of distribution; Variance ($\sigma^2$) shows how spread out the data is.

```python
# Simulating 10,000 coin tosses (1 = Heads, 0 = Tails)
np.random.seed(42)
tosses = np.random.choice([0, 1], size=10000, p=[0.5, 0.5])
prob_heads = np.mean(tosses == 1)
print(f"Estimated Probability of Heads: {prob_heads:.4f}")

# Bayes' Theorem Example: Spam Filter
# P(Spam) = 0.2, P(Free | Spam) = 0.8, P(Free | Not Spam) = 0.1
p_spam = 0.20
p_not_spam = 0.80
p_free_given_spam = 0.80
p_free_given_not_spam=0.10
p_free = (p_free_given_spam * p_spam) + (p_free_given_not_spam * p_not_spam)

p_spam_given_free = (p_free_given_spam * p_spam) / p_free
print(f"Probability email is Spam given it contains 'Free': {p_spam_given_free:.2%}")
```

## 4. Gradient Intuition (Gradient Descent)
- **What is a Gradient?** A gradient is a vector of partial derivatives that points in the direction of the **steepest increase** of a function.
- **Intuition (Walking Downhill Blindfolded):** Imagine standing on a foggy hill and wanting to reach the lowest valley (minimum loss/error). You feel the slope under your feet and take a step in the **downhill (negative gradient)** direction.
- **Gradient Descent Update Rule:**
  $$\theta_{\text{new}} = \theta_{\text{old}} - \alpha \cdot \nabla J(\theta)$$
  - $\theta$: Model parameter (weight)
  - $\alpha$: Learning rate (step size — too big overshoots the valley; too small takes too long)
  - $\nabla J(\theta)$: Gradient (slope of the loss function)

```python
# Finding the minimum of loss function L(w) = (w - 3)^2 using Gradient Descent
# Derivative (gradient) of L(w) is: dL/dw = 2 * (w - 3)

w = 0.0           # Starting guess
learning_rate = 0.1

for step in range(1, 16):
    loss = (w - 3) ** 2
    grad = 2 * (w - 3)
    w = w - learning_rate * grad  # Move opposite to gradient
    if step in [1, 5, 10, 15]:
        print(f"Step {step:2d} | w = {w:.4f} | Loss = {loss:.6f}")
```
